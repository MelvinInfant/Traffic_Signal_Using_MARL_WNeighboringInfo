```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as steps_range
```

```python
df=pd.read_csv("/content/baseline_stress_log.csv")
df1=pd.read_csv("/content/proposed_stress_log.csv")
plt.figure(figsize=(20,15))
plt.plot(df['Episode'], df['Total_Reward'],label='CTDE')
plt.plot(df1['Episode'], df1['Total_Reward'],label='CTDE_N')
plt.xlabel('Episode')
plt.ylabel('Total_Reward')
plt.legend()
plt.show()
```

```python

common_columns = [col for col in df.columns if col in df1.columns and col != 'Episode']

for col in common_columns:
    plt.figure(figsize=(10, 6))
    plt.plot(df['Episode'], df[col], label=f'CTDE - {col}')
    plt.plot(df1['Episode'], df1[col], label=f'CTDE_N - {col}')
    plt.xlabel('Episode')
    plt.ylabel(col)
    plt.title(f'Comparison of {col} between df and df1')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
print('Variance of df Total_Reward:', df['Total_Reward'].var())
print('Mean of df Total_Reward:', df['Total_Reward'].mean())
print('Variance of df Steps:', df['Steps'].var())
print('Variance of df1 Total_Reward:', df1['Total_Reward'].var())
print('Mean of df1 Total_Reward:', df1['Total_Reward'].mean())
print('Variance of df1 Steps:', df1['Steps'].var())
```

### Comparison of Variance and Mean for Total_Reward

```python
import numpy as np

df_tr_var = df['Total_Reward'].var()
df_tr_mean = df['Total_Reward'].mean()
df1_tr_var = df1['Total_Reward'].var()
df1_tr_mean = df1['Total_Reward'].mean()

labels = ['Baseline Total_Reward', 'Hypothesis Total_Reward']
variance_values = [df_tr_var, df1_tr_var]
mean_values = [df_tr_mean, df1_tr_mean]

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, variance_values, width, label='Variance')
rects2 = ax.bar(x + width/2, mean_values, width, label='Mean')

ax.set_ylabel('Value')
ax.set_title('Comparison of Variance and Mean for Total_Reward')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()

def autolabel(rects):
    """Attach a text label above each bar in *rects*, displaying its height."""
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)

fig.tight_layout()
plt.show()
```
